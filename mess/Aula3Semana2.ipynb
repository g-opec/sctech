```python
quant_aluno = int(input("Quantidade de alunos a ser cadastrados "))
cad_aluno = 0
aprovado = 0
recuperacao = 0
reprovado = 0
maiormedia = 0
somamedia = 0

while cad_aluno < quant_aluno:
  nome_aluno=input("Nome do aluno ")
  idade_aluno=int(input("Idade do aluno"))
  nota1=float(input("Nota 1 "))
  nota2=float(input("Nota 2 "))
  nota3=float(input("Nota 3 "))
  if nota1 < 0 or nota2 < 0 or nota3 < 0:
    continue
  media = (nota1+nota2+nota3)/3
  if media >= 7:
    final = "Aprovado"
    aprovado += 1
  elif media >= 5 and media < 7:
    final = "Recuperação"
    recuperacao += 1
  elif media < 5:
    final = "Reprovado"
    reprovado += 1
  if idade_aluno < 15 and media >= 7:
    print(f"Aluno {nome_aluno} destaque jovem!")
  if media > maiormedia:
    maior = [nome_aluno, media]
  somamedia += media
  cad_aluno += 1

mediageral = somamedia/quant_aluno
print(f"Total de alunos cadastrados {quant_aluno}.")
print(f"Quantidade de aprovados {aprovado}")
print(f"Quantidade em recuperação {recuperacao}")
print(f"Quantidade de reprovados {reprovado}")
print(f"Maior média foi de {maior[1]} do(a) aluno(a) {maior[0]}")
print(f"Média geral da turma {mediageral}")
if mediageral < 6:
  print(f"Atenção! Nota média da turma abaixo de 6")
```
